# Unveil — Sketch → 3D (Google Colab)

Draw or upload a sketch, and Unveil turns it into a textured 3D model (GLB/OBJ):
**sketch cleanup → SD1.5 + ControlNet (3–4 creativity levels, auto-selected) → TripoSR multi-candidate reconstruction → silhouette verification → depth control → export.**

**Before you start:** `Runtime → Change runtime type → T4 GPU` (the free tier works).

Run the cells top to bottom. The first generation downloads about 6 GB of model weights (3–6 min on Colab). Later runs take about 1–2 min.
